```python
# CODE FÜR MBERT

!pip -q install transformers sentencepiece requests

import re, requests, pandas as pd
from transformers import pipeline

fill = pipeline("fill-mask", model="bert-base-multilingual-cased")

titles = ["Berlin","Water","Earth","Photosynthesis","Machine_learning",
          "Python_(programming_language)","Albert_Einstein","World_War_II",
          "Solar_System","DNA"]

STOP = set("the a an and or but if then this that these those is are was were be been being of to in on for with as by from at into".split())

def wiki_first_sentence(title):
    url = f"https://en.wikipedia.org/api/rest_v1/page/summary/{title}"
    r = requests.get(url, headers={"User-Agent":"kaggle-mbert-enwiki/1.0"}, timeout=30)
    if r.status_code != 200:
        return None, None, None, f"http_{r.status_code}"
    j = r.json()
    if "extract" not in j or "content_urls" not in j or not j["extract"].strip():
        return None, None, None, "no_extract"
    sent = re.split(r"(?<=[.!?])\s+", j["extract"].strip())[0]
    rev = j.get("revision", None)
    page_url = j["content_urls"]["desktop"]["page"]
    return sent, rev, page_url, ""

def make_mask(sent):
    for w in re.findall(r"[A-Za-z]+", sent):
        wl = w.lower()
        if wl in STOP or len(wl) < 4:
            continue
        # single-token constraint for THIS model (mBERT)
        if len(fill.tokenizer.tokenize(wl)) != 1:
            continue
        masked = re.sub(rf"\b{re.escape(w)}\b", "[MASK]", sent, count=1)
        return wl, masked
    return None, None

rows = []
for t in titles:
    sent, rev, url, err = wiki_first_sentence(t)

    if err:
        rows.append({
            "title": t, "revision": rev, "url": url,
            "masked_sentence": "", "gold": "",
            "top1": "", "top5": "",
            "top1_correct": False, "top5_correct": False,
            "error": err
        })
        continue

    gold, masked = make_mask(sent)
    if not gold:
        rows.append({
            "title": t, "revision": rev, "url": url,
            "masked_sentence": sent, "gold": "",
            "top1": "", "top5": "",
            "top1_correct": False, "top5_correct": False,
            "error": "no_single_token_word_found"
        })
        continue

    preds = fill(masked)[:5]
    top5 = [p["token_str"].strip().lower() for p in preds]
    top1 = top5[0] if top5 else ""

    rows.append({
        "title": t, "revision": rev, "url": url,
        "masked_sentence": masked, "gold": gold,
        "top1": top1, "top5": ", ".join(top5),
        "top1_correct": (top1 == gold),
        "top5_correct": (gold in top5),
        "error": ""
    })

df = pd.DataFrame(rows)
display(df[["title","revision","url","masked_sentence","gold","top1","top5_correct","error"]])

print("mBERT – Wikipedia (EN)")
print("Top-1:", int(df.top1_correct.sum()), "/", len(df))
print("Top-5:", int(df.top5_correct.sum()), "/", len(df))
```

```python
# CODE FÜR HUMAN
!pip -q install transformers sentencepiece requests pandas

import re, requests, pandas as pd
from transformers import AutoTokenizer

# Use ALBERT tokenizer ONLY so the masked word selection is EXACTLY like your ALBERT code
tok = AutoTokenizer.from_pretrained("albert-base-v2")

titles = ["Berlin","Water","Earth","Photosynthesis","Machine_learning",
          "Python_(programming_language)","Albert_Einstein","World_War_II",
          "Solar_System","DNA"]

STOP = set("the a an and or but if then this that these those is are was were be been being of to in on for with as by from at into".split())

def wiki_first_sentence(title):
    j = requests.get(
        f"https://en.wikipedia.org/api/rest_v1/page/summary/{title}",
        headers={"User-Agent":"kaggle-human-wiki/1.0"},
        timeout=30
    ).json()
    sent = re.split(r"(?<=[.!?])\s+", j["extract"].strip())[0]
    return sent, j.get("revision"), j["content_urls"]["desktop"]["page"]

def make_mask_same_as_albert(sent):
    # SAME rule as your ALBERT code:
    # - skip stopwords/short words
    # - only accept single-token words under ALBERT tokenizer
    for w in re.findall(r"[A-Za-z]+", sent):
        wl = w.lower()
        if wl in STOP or len(wl) < 4:
            continue
        if len(tok.tokenize(wl)) != 1:
            continue
        masked = re.sub(rf"\b{re.escape(w)}\b", "[MASK]", sent, count=1)
        return wl, masked
    return None, None

def norm_word(x):
    # robust: compare only letters
    return re.sub(r"[^a-z]", "", str(x).lower().strip())

# 1) Build the SAME 10 masked sentences + gold labels
items = []
for t in titles:
    try:
        sent, rev, url = wiki_first_sentence(t)
        gold, masked = make_mask_same_as_albert(sent)

        if not gold:
            items.append({
                "title": t, "revision": rev, "url": url,
                "masked_sentence": "", "gold": "", "error": "no_single_token_word_found"
            })
        else:
            items.append({
                "title": t, "revision": rev, "url": url,
                "masked_sentence": masked, "gold": gold, "error": ""
            })

    except Exception as e:
        items.append({
            "title": t, "revision": None, "url": "",
            "masked_sentence": "", "gold": "", "error": repr(e)
        })

df_items = pd.DataFrame(items)

# Show prompts for humans (DON'T show gold when collecting answers)
display(df_items[["title","revision","url","masked_sentence","error"]])

# 2) HUMAN answers
# Set MODE="demo" once to verify scoring works (it will fill top1 with gold)
MODE = "demo"   # change to "manual" when you enter real answers

human_answers = [
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
]

# 3) Score humans (always /10)
rows = []
for i, row in df_items.iterrows():
    gold = norm_word(row["gold"])

    if MODE == "demo":
        top1 = gold
        top5 = [gold]
    else:
        top1 = norm_word(human_answers[i].get("human_top1",""))
        top5 = [norm_word(x) for x in human_answers[i].get("human_top5", [])]
        top5 = [x for x in top5 if x]

    if gold == "" or row["masked_sentence"] == "":
        rows.append({**row,
                     "human_top1": top1,
                     "human_top5": ", ".join(top5),
                     "top1_correct": False,
                     "top5_correct": False})
        continue

    rows.append({**row,
                 "human_top1": top1,
                 "human_top5": ", ".join(top5),
                 "top1_correct": (top1 == gold),
                 "top5_correct": (gold in top5) if top5 else False})

df_human = pd.DataFrame(rows)
display(df_human[["title","url","gold","human_top1","top1_correct","top5_correct","error"]])

print("\nHUMAN – Wikipedia (EN) – same sentences as ALBERT setup")
print("Top-1:", int(df_human.top1_correct.sum()), "/", len(df_human))
print("Top-5:", int(df_human.top5_correct.sum()), "/", len(df_human))
```

```python
# CODE FÜR BERT

!pip -q install transformers sentencepiece requests beautifulsoup4 lxml

import re, requests, pandas as pd
from bs4 import BeautifulSoup
from transformers import pipeline

fill = pipeline("fill-mask", model="bert-base-german-cased")

# 10 echte deutsche Texte auf Wikisource (Grimm, Public Domain)
pages = [
    "Rothkäppchen_(1812)",
    "Aschenputtel_(1812)",
    "Hänsel_und_Gretel_(1812)",
    "Rumpelstilzchen_(1812)",
    "Sneewittchen_(Schneeweißchen)_(1812)",
    "Rothkäppchen_(1837)",
    "Aschenputtel_(1857)",
    "Rumpelstilzchen_(1819)",
    "Sneewittchen_(1857)",
    "Hänsel_und_Grethel_(1840)",
]

BASE = "https://de.wikisource.org"
API  = f"{BASE}/w/api.php"

STOP = set("der die das und oder aber wenn dann ist sind war waren von mit auf in für als bei über unter im am an des dem den".split())

def get_plain_text_from_wikisource(title):
    # MediaWiki Action API: parse -> HTML; then strip tags
    params = {
        "action": "parse",
        "page": title,
        "prop": "text",
        "format": "json",
        "formatversion": "2",
    }
    r = requests.get(API, params=params, headers={"User-Agent":"kaggle-de-bert/1.0"}, timeout=25)
    r.raise_for_status()
    j = r.json()

    html = j["parse"]["text"]
    rev  = j["parse"].get("revid", None)
    url  = f"{BASE}/wiki/{title}"

    soup = BeautifulSoup(html, "lxml")
    text = soup.get_text(" ", strip=True)
    text = re.sub(r"\s+", " ", text)
    return text, rev, url

def first_good_sentence(text):
    sents = [s.strip() for s in re.split(r"(?<=[.!?])\s+", text) if len(s.strip()) > 50]
    return sents[0] if sents else None

def make_mask(sent):
    for w in re.findall(r"[A-Za-zÄÖÜäöüß]+", sent):
        wl = w.lower()
        if wl in STOP or len(wl) < 4:
            continue
        if len(fill.tokenizer.tokenize(wl)) != 1:   # fairness: single-token target
            continue
        masked = re.sub(rf"\b{re.escape(w)}\b", "[MASK]", sent, count=1)
        return wl, masked
    return None, None

rows = []
skipped = []

for title in pages:
    try:
        text, rev, url = get_plain_text_from_wikisource(title)
        sent = first_good_sentence(text)
        if not sent:
            skipped.append((title, "no_sentence"))
            continue

        gold, masked = make_mask(sent)
        if gold is None:
            skipped.append((title, "no_single_token_word_found"))
            continue

        preds = fill(masked)[:5]
        top5 = [p["token_str"].strip().lower() for p in preds]
        rows.append({
            "title": title, "revision": rev, "url": url,
            "masked_sentence": masked, "gold": gold,
            "top1": top5[0], "top5": ", ".join(top5),
            "top1_correct": top5[0] == gold,
            "top5_correct": gold in top5
        })
    except Exception as e:
        skipped.append((title, repr(e)))

df = pd.DataFrame(rows)
display(df[["title","revision","url","masked_sentence","gold","top1","top5_correct"]])

print("German BERT – Deutsche Texte (Wikisource)")
print("Built items:", len(df), "/", len(pages))
print("Top-1:", int(df.top1_correct.sum()), "/", len(df))
print("Top-5:", int(df.top5_correct.sum()), "/", len(df))

print("\nSkipped (falls welche):")
for s in skipped:
    print("-", s)
```

```python
# CODE FÜR HUMAN
!pip -q install requests beautifulsoup4 lxml pandas

import re, requests, pandas as pd
from bs4 import BeautifulSoup

pages = [
    "Rothkäppchen_(1812)",
    "Aschenputtel_(1812)",
    "Hänsel_und_Gretel_(1812)",
    "Rumpelstilzchen_(1812)",
    "Sneewittchen_(1812)",
    "Frau_Holle_(1812)",
    "Der_Froschkönig_oder_der_eiserne_Heinrich_(1812)",
    "Die_sieben_Raben_(1812)",
    "Dornröschen_(1812)",
    "Der_Tapfere_Schneiderlein_(1812)",
]

BASE = "https://de.wikisource.org"
API  = f"{BASE}/w/api.php"

STOP = set("""
der die das und oder aber wenn dann ist sind war waren von mit auf in für als bei über unter im am an des dem den
ein eine einer eines einem einen
ich du er sie es wir ihr ihnen ihnen mich dich sich
""".split())

def get_plain_text(title):
    params = {"action":"parse","page":title,"prop":"text","format":"json","formatversion":"2"}
    r = requests.get(API, params=params, headers={"User-Agent":"kaggle-human-wikisource/1.0"}, timeout=30)
    r.raise_for_status()
    j = r.json()

    html = j["parse"]["text"]
    rev  = j["parse"].get("revid", None)
    url  = f"{BASE}/wiki/{title}"

    soup = BeautifulSoup(html, "lxml")
    text = re.sub(r"\s+", " ", soup.get_text(" ", strip=True))
    return text, rev, url

def pick_sentences(text, max_sents=60):
    sents = [s.strip() for s in re.split(r"(?<=[.!?])\s+", text) if len(s.strip()) > 60]
    return sents[:max_sents]

def make_mask_human(sent):
    # Humans don't need tokenizer constraints; pick a reasonable content word
    # Avoid super-short/stopwords; mask first suitable word
    for w in re.findall(r"[A-Za-zÄÖÜäöüß]+", sent):
        wl = w.lower()
        if wl in STOP or len(wl) < 5:
            continue
        masked = re.sub(rf"\b{re.escape(w)}\b", "[MASK]", sent, count=1)
        return wl, masked
    return None, None

def norm_word_de(x):
    # keep only letters incl. German umlauts; lower-case
    return re.sub(r"[^a-zäöüß]", "", str(x).lower().strip())

# -----------------------------
# 1) Build 10 German prompts
# -----------------------------
items = []
for title in pages:
    rev, url = None, f"{BASE}/wiki/{title}"
    try:
        text, rev, url = get_plain_text(title)

        chosen = None
        for s in pick_sentences(text, max_sents=60):
            gold, masked = make_mask_human(s)
            if gold:
                chosen = (gold, masked)
                break

        if chosen is None:
            items.append({
                "title": title, "revision": rev, "url": url,
                "masked_sentence": "", "gold": "", "error": "no_maskable_sentence_found"
            })
        else:
            gold, masked = chosen
            items.append({
                "title": title, "revision": rev, "url": url,
                "masked_sentence": masked, "gold": gold, "error": ""
            })

    except Exception as e:
        items.append({
            "title": title, "revision": rev, "url": url,
            "masked_sentence": "", "gold": "", "error": repr(e)
        })

df_items = pd.DataFrame(items)

# Show prompts for humans (don't show gold when collecting answers!)
display(df_items[["title","revision","url","masked_sentence","error"]])

# -----------------------------
# 2) HUMAN answers
# -----------------------------
# "demo"  -> fills human_top1 with gold so you immediately see scoring works
# "manual"-> you fill the answers below
MODE = "demo"

human_answers = [
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
]

# -----------------------------
# 3) Score humans (always /10)
# -----------------------------
rows = []
for i, row in df_items.iterrows():
    gold = norm_word_de(row["gold"])

    if MODE == "demo":
        top1 = gold
        top5 = [gold]
    else:
        top1 = norm_word_de(human_answers[i].get("human_top1",""))
        top5 = [norm_word_de(x) for x in human_answers[i].get("human_top5", [])]
        top5 = [x for x in top5 if x]

    # keep denominator 10
    if gold == "" or row["masked_sentence"] == "":
        rows.append({**row,
                     "human_top1": top1,
                     "human_top5": ", ".join(top5),
                     "top1_correct": False,
                     "top5_correct": False})
        continue

    rows.append({**row,
                 "human_top1": top1,
                 "human_top5": ", ".join(top5),
                 "top1_correct": (top1 == gold),
                 "top5_correct": (gold in top5) if top5 else False})

df_human = pd.DataFrame(rows)
display(df_human[["title","url","gold","human_top1","top1_correct","top5_correct","error"]])

print("\nHUMAN baseline – German texts (Wikisource)")
print("Top-1:", int(df_human.top1_correct.sum()), "/", len(df_human))
print("Top-5:", int(df_human.top5_correct.sum()), "/", len(df_human))
```

```python
# CODE FÜR HUMAN
!pip -q install requests pandas

import re, requests, pandas as pd

books = [
    ("Pride and Prejudice", "Jane Austen", "https://www.gutenberg.org/cache/epub/1342/pg1342.txt"),
    ("Alice's Adventures in Wonderland", "Lewis Carroll", "https://www.gutenberg.org/cache/epub/11/pg11.txt"),
    ("Peter Pan", "J. M. Barrie", "https://www.gutenberg.org/cache/epub/16/pg16.txt"),
    ("The Adventures of Sherlock Holmes", "Arthur Conan Doyle", "https://www.gutenberg.org/cache/epub/1661/pg1661.txt"),
    ("The Time Machine", "H. G. Wells", "https://www.gutenberg.org/cache/epub/35/pg35.txt"),
    ("Dracula", "Bram Stoker", "https://www.gutenberg.org/cache/epub/345/pg345.txt"),
    ("Frankenstein", "Mary Shelley", "https://www.gutenberg.org/cache/epub/84/pg84.txt"),
    ("A Tale of Two Cities", "Charles Dickens", "https://www.gutenberg.org/cache/epub/98/pg98.txt"),
    ("The Wonderful Wizard of Oz", "L. Frank Baum", "https://www.gutenberg.org/cache/epub/55/pg55.txt"),
    ("Moby-Dick", "Herman Melville", "https://www.gutenberg.org/cache/epub/2701/pg2701.txt"),
]

STOP = set("the a an and or but if then this that these those is are was were be been being of to in on for with as by from at into".split())

# ---------- Better Gutenberg cleaning ----------
def clean_gutenberg_text(txt: str) -> str:
    # Most reliable: strip header/footer using the *** START/*** END markers
    start = re.search(r"\*\*\*\s*START OF (THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*", txt, flags=re.IGNORECASE|re.DOTALL)
    end   = re.search(r"\*\*\*\s*END OF (THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*", txt, flags=re.IGNORECASE|re.DOTALL)
    if start and end and start.end() < end.start():
        txt = txt[start.end():end.start()]

    # remove common Gutenberg noise lines
    txt = txt.replace("\r", "\n")
    txt = re.sub(r"\n{2,}", "\n", txt)
    txt = re.sub(r"[ \t]+", " ", txt)

    # remove bracketed illustration markers etc.
    txt = re.sub(r"\[\s*Illustration\s*\]", " ", txt, flags=re.IGNORECASE)
    txt = re.sub(r"\[\s*Illustrations?\s*\]", " ", txt, flags=re.IGNORECASE)

    return txt.strip()

def get_sentences(url):
    raw = requests.get(url, headers={"User-Agent":"kaggle-human-baseline/1.0"}, timeout=60).text
    txt = clean_gutenberg_text(raw)
    # sentence split
    sents = [s.strip() for s in re.split(r"(?<=[.!?])\s+", txt) if len(s.strip()) > 40]
    return sents

def choose_mask(sentence):
    # pick a content-ish word; (for humans we don't need tokenizer constraints)
    for w in re.findall(r"[A-Za-z]+", sentence):
        wl = w.lower()
        if wl in STOP or len(wl) < 5:
            continue
        masked = re.sub(rf"\b{re.escape(w)}\b", "[MASK]", sentence, count=1)
        return wl, masked
    return None, None

items = []
for title, author, url in books:
    sents = get_sentences(url)
    chosen = None

    # 5-sentence window, mask in middle sentence
    for i in range(2, min(len(sents)-2, 400)):
        gold, masked_mid = choose_mask(sents[i])
        if not gold:
            continue
        long_text = " ".join(sents[i-2:i] + [masked_mid] + sents[i+1:i+3])
        if 250 <= len(long_text) <= 900:  # keep it reasonably "long"
            chosen = (gold, long_text)
            break

    if chosen is None:
        items.append({"book": title, "author": author, "source_url": url,
                      "gold": "", "masked_long_text": "", "note": "could_not_build_item"})
    else:
        gold, long_text = chosen
        items.append({"book": title, "author": author, "source_url": url,
                      "gold": gold, "masked_long_text": long_text, "note": ""})

df_items = pd.DataFrame(items)


display(df_items[["book","author","source_url","masked_long_text"]])



# Fill in ONLY if HUMAN_MODE="manual"
human_answers = [
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
    {"human_top1":"", "human_top5":[]},
]

rows = []
for i, row in df_items.iterrows():
    gold = str(row["gold"]).strip().lower()

    if HUMAN_MODE == "demo":
        top1 = gold
        top5 = [gold]
    else:
        ans = human_answers[i]
        top1 = str(ans.get("human_top1","")).strip().lower()
        top5 = [str(x).strip().lower() for x in ans.get("human_top5", [])]

    # count always /10
    if gold == "" or row["masked_long_text"] == "":
        rows.append({**row, "human_top1": top1, "human_top5": ", ".join(top5),
                     "top1_correct": False, "top5_correct": False})
        continue

    rows.append({**row,
        "human_top1": top1,
        "human_top5": ", ".join(top5),
        "top1_correct": (top1 == gold),
        "top5_correct": (gold in top5) if top5 else False
    })

df_human = pd.DataFrame(rows)
display(df_human[["book","source_url","gold","human_top1","top1_correct","top5_correct","note"]])

print("\nHUMAN baseline – LONG CONTEXT (5 sentences)")
print("Top-1:", int(df_human.top1_correct.sum()), "/", len(df_human))
print("Top-5:", int(df_human.top5_correct.sum()), "/", len(df_human))

```
